```python
import pandas as pd
from IPython.display import display

df = pd.read_csv("../results/phase2/2a/sparse_dense_results.csv")

# Table 1: Memory consumption
memory_table = df[[
    "corpus_size",
    "sparse_mem_mb",
    "dense_mem_mb",
    "reduced_mem_mb"
]].copy()

memory_table.columns = [
    "Corpus Size",
    "Sparse TF-IDF (MB)",
    "Dense TF-IDF (MB)",
    "Reduced SVD (MB)"
]

display(memory_table.round(2))

# Table 2: Computation time and variability
time_columns = [
    "sparse_sparse_scipy_time",
    "sparse_sparse_scipy_time_std_seconds",
    "sparse_sparse_manual_time",
    "sparse_sparse_manual_time_std_seconds",
    "sparse_dense_time",
    "sparse_dense_time_std_seconds",
    "dense_dense_time",
    "dense_dense_time_std_seconds"
]

time_table = df[["corpus_size"] + time_columns].copy()

time_table.columns = [
    "Corpus Size",
    "Sparse-Sparse SciPy Mean (s)",
    "Sparse-Sparse SciPy Std (s)",
    "Sparse-Sparse Manual Mean (s)",
    "Sparse-Sparse Manual Std (s)",
    "Sparse-Dense Mean (s)",
    "Sparse-Dense Std (s)",
    "Dense-Dense Mean (s)",
    "Dense-Dense Std (s)"
]

display(time_table.round(4))
```
